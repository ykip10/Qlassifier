```python
from Qlassifier import api
```

```python
# Set the scope: popular VCE subjects with large exam component
subjects = [
    "specialist_mathematics",
    "mathematical_methods",
    "physics",
    "chemistry",
    "business_management", 
]

years = range(2017, 2024) # Set timeframe

all_exams = api.process_material(subjects, years, type="exam")
all_reports = api.process_material(subjects, years, type="report")
all_sds = api.process_material(subjects, years, type="study_design")

assert(all_exams and all_reports and all_sds)
```
